```python
#constants
me = 0.000511
mmu = 0.1057
mp = 0.9383

beam_energies = [0.5, 1.0, 2.0, 10.0, 100.0]
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def calc_T_max(E, mass):
    gamma = E/mass
    print(E, mass)
    beta = np.sqrt(1-1/gamma**2)

    return (2.0 * me * (beta*gamma)**2)/(1+(2*gamma*me/mass +(me/mass)**2))
```

```python
def calculate_angle(beam_energies, mass):
    T_results = [[None]*1000]*len(beam_energies)
    angle_results = [[None]*1000]*len(beam_energies)

    for E, energy in enumerate(beam_energies):

        T_max = calc_T_max(energy, mass)

        T_vals = np.linspace(1e-12, 0.999*T_max, 1000)

        thetas = np.arccos(np.sqrt(T_vals/T_max))

        angle_results[E] = thetas
        T_results[E] = T_vals


    return T_results, angle_results
```

```python
mu_Ts, mu_angles = calculate_angle(beam_energies, mmu)
```

```python
fig, ax = plt.subplots()
facecolors = ["Blue", "Green", "Red", "Black", "Orange"]

for i in range(len(facecolors)):
    ax.plot(mu_Ts[i], mu_angles[i], color=facecolors[i], label=f"$\epsilon$={beam_energies[i]} GeV")
    ax.set_xscale("log")

ax.legend()
ax.set_xlabel("Energy (GeV)")
ax.set_ylabel("Angle")
ax.set_title(r"$\delta$-electron emission angle as function of its energy from a muon")
```

```python
p_Ts, p_angles = calculate_angle(beam_energies, mp)
```

```python
fig, ax = plt.subplots()
facecolors = ["Blue", "Green", "Red", "Black", "Orange"]

for i in range(len(facecolors)):
    ax.plot(p_Ts[i], p_angles[i], color=facecolors[i], label=f"$\epsilon$={beam_energies[i]} GeV")
    ax.set_xscale("log")

ax.legend()
ax.set_xlabel("Energy (GeV)")
ax.set_ylabel("Angle")
ax.set_title(r"$\delta$-electron emission angle as function of its energy from a proton")
```
